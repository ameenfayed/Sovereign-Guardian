```python
# ============================================================
# VISUALIZATION 3: DIGITAL TWIN — ANNUAL PERFORMANCE SIMULATION
# ============================================================

# Gulf climate data (Hail, Saudi Arabia - annual)
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
days_in_month = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

# Monthly averages for Hail, Saudi Arabia
temp_c = [13.5, 16.2, 20.8, 26.4, 33.1, 38.9, 41.2, 40.8, 36.4, 28.7, 20.3, 14.8]
rh_pct = [35, 28, 22, 18, 13, 11, 12, 14, 17, 22, 31, 38]
uv_wm2 = [420, 560, 720, 880, 980, 1050, 1040, 990, 870, 700, 510, 390]

# MIRAG yield model: base yield + corrections
# Yield (L/day/module) = f(RH, T, UV, atmospheric E-field)
def mirag_yield(rh, temp, uv):
    """MIRAG™ daily yield model based on physical parameters."""
    # QD sorption capacity (L/day) — increases with lower T (higher ΔT for condensation)
    # and with RH (more vapour available)
    rh_factor = max(0, (rh - 10) / 40)  # normalized, 0 at 10% RH, 1 at 50%
    t_factor = max(0.3, 1.0 - (temp - 20) * 0.012)  # condensation efficiency decreases at high T
    uv_factor = min(1.2, uv / 800)  # UV boosts TiO2 photocatalysis & PVDF power
    inn17_boost = 1.15 if rh < 15 else 1.0  # Inn.17 corona enhancement at very low RH
    inn20_boost = 1.20 if rh < 15 else 1.05  # Inn.20 atmospheric E-field boost (more pronounced at low RH)
    
    base_yield = 80 * rh_factor * t_factor  # L/day at standard conditions
    return base_yield * uv_factor * inn17_boost * inn20_boost

yields_daily = [mirag_yield(r, t, u) for r, t, u in zip(rh_pct, temp_c, uv_wm2)]
yields_monthly = [y * d for y, d in zip(yields_daily, days_in_month)]
yields_competitor_SOURCE = [max(0, (r - 25) * 1.2) for r in rh_pct]  # SOURCE Hydropanel needs >25% RH
yields_competitor_SOURCE_monthly = [y * d for y, d in zip(yields_competitor_SOURCE, days_in_month)]

# CO2 capture monthly (proportional to yield + UV)
co2_monthly = [(u/1000 * 0.3 + 0.25) * d for u, d in zip(uv_wm2, days_in_month)]  # kg CO2

# Carbon credit revenue (USD, at $100/ton Verra VCS)
carbon_revenue_monthly = [c * 0.1 for c in co2_monthly]  # $100/ton = $0.1/kg

fig_twin = make_subplots(
    rows=3, cols=2,
    subplot_titles=[
        '💧 Monthly Water Yield (L) — MIRAG vs Competitors',
        '🌡️ Temperature & RH — Gulf Desert Annual Profile',
        '☀️ UV Irradiance (W/m²)',
        '🌱 CO₂ Capture (kg/month)',
        '💰 Carbon Credit Revenue (USD/month)',
        '📈 Cumulative Annual Water Production (L)'
    ],
    vertical_spacing=0.12,
    horizontal_spacing=0.10
)

# Row 1, Col 1: Yield comparison
fig_twin.add_trace(go.Bar(
    x=months, y=yields_monthly,
    name='MIRAG™ (with Inn.17+18+20)',
    marker=dict(color=[MIRAG_GREEN if y > 0 else '#334455' for y in yields_monthly],
                line=dict(color=MIRAG_DARK, width=0.5)),
), row=1, col=1)

fig_twin.add_trace(go.Bar(
    x=months, y=yields_competitor_SOURCE_monthly,
    name='SOURCE Hydropanel (>25% RH only)',
    marker=dict(color=MIRAG_RED, opacity=0.7),
), row=1, col=1)

# Row 1, Col 2: Temperature & RH
fig_twin.add_trace(go.Scatter(
    x=months, y=temp_c, name='Temperature (°C)',
    line=dict(color=MIRAG_ORANGE, width=2.5),
    fill='tozeroy', fillcolor='rgba(255,107,53,0.08)'
), row=1, col=2)

fig_twin.add_trace(go.Scatter(
    x=months, y=rh_pct, name='Humidity (%)',
    line=dict(color=MIRAG_BLUE, width=2.5),
    yaxis='y5'
), row=1, col=2)

fig_twin.add_hrect(y0=10, y1=15, fillcolor='rgba(255,51,85,0.1)',
                   line=dict(width=0), row=1, col=2,
                   annotation_text='Critical Zone', annotation_font=dict(color=MIRAG_RED, size=9))

# Row 2, Col 1: UV
fig_twin.add_trace(go.Scatter(
    x=months, y=uv_wm2, name='UV Irradiance',
    line=dict(color=MIRAG_GOLD, width=2.5),
    fill='tozeroy', fillcolor='rgba(255,215,0,0.08)'
), row=2, col=1)

# Row 2, Col 2: CO2
fig_twin.add_trace(go.Bar(
    x=months, y=co2_monthly, name='CO₂ Captured',
    marker=dict(color=MIRAG_GREEN, opacity=0.8)
), row=2, col=2)

# Row 3, Col 1: Carbon Revenue
fig_twin.add_trace(go.Bar(
    x=months, y=carbon_revenue_monthly, name='Carbon Credits USD',
    marker=dict(color=MIRAG_PURPLE, opacity=0.8)
), row=3, col=1)

# Row 3, Col 2: Cumulative water
cum_mirag = np.cumsum(yields_monthly)
cum_source = np.cumsum(yields_competitor_SOURCE_monthly)
fig_twin.add_trace(go.Scatter(
    x=months, y=cum_mirag, name='MIRAG™ Cumulative',
    line=dict(color=MIRAG_GREEN, width=3),
    fill='tozeroy', fillcolor='rgba(0,255,170,0.08)'
), row=3, col=2)
fig_twin.add_trace(go.Scatter(
    x=months, y=cum_source, name='SOURCE Cumulative',
    line=dict(color=MIRAG_RED, width=2, dash='dash')
), row=3, col=2)

fig_twin.update_layout(
    height=900,
    title=dict(
        text='🖥️ MIRAG™ Digital Twin — Annual Performance Simulation (Hail, Saudi Arabia)',
        font=dict(color=MIRAG_BLUE, size=18, family='Courier New')
    ),
    **PLOTLY_DARK,
    legend=dict(bgcolor='#0a1628', bordercolor='#223344', borderwidth=1,
                font=dict(color='#aabbcc', size=10), x=0.01, y=0.99)
)

fig_twin.show()

total_annual_yield = sum(yields_monthly)
total_co2 = sum(co2_monthly) / 1000  # tons
total_carbon_revenue = sum(carbon_revenue_monthly)
source_annual = sum(yields_competitor_SOURCE_monthly)

print(f"\n{'='*60}")
print(f"📊 MIRAG™ DIGITAL TWIN — ANNUAL SUMMARY (Hail, KSA)")
print(f"{'='*60}")
print(f"💧 Total Water Yield:         {total_annual_yield:.0f} L/yr")
print(f"🌱 CO₂ Captured:              {total_co2:.2f} tons/yr")
print(f"💰 Carbon Credit Revenue:     USD {total_carbon_revenue:.0f}/yr")
print(f"🏆 vs SOURCE Hydropanel:      {total_annual_yield/max(source_annual,1):.1f}× more water")
print(f"   (SOURCE fails Jun-Oct at <25% RH — MIRAG operates all year)")
```

```python
# ============================================================
# VISUALIZATION 6: SELF-EVOLVING SYSTEM — INN.19+22 SIMULATION
# The Membrane That Learns From Itself
# ============================================================

# Simulate Inn.22: Self-Replicating Parameter Protocol
# Each generation learns from the previous fleet's performance data
np.random.seed(2026)

generations = np.arange(1, 9)  # Generation 1 (2027) to 8 (2034)

# Performance metrics improving each generation
# Gen 1 baseline, each subsequent gen improves by 8-15%
yield_efficiency = [1.00]
rh_min_threshold = [15.0]
self_healing_cycles = [10000]
co2_per_unit = [6.5]
pfas_removal_pct = [96.0]

for g in range(1, 8):
    yield_efficiency.append(yield_efficiency[-1] * (1 + np.random.uniform(0.07, 0.14)))
    rh_min_threshold.append(max(8.0, rh_min_threshold[-1] * (1 - np.random.uniform(0.04, 0.10))))
    self_healing_cycles.append(self_healing_cycles[-1] * (1 + np.random.uniform(0.08, 0.15)))
    co2_per_unit.append(co2_per_unit[-1] * (1 + np.random.uniform(0.05, 0.12)))
    pfas_removal_pct.append(min(99.9, pfas_removal_pct[-1] * (1 + np.random.uniform(0.003, 0.008))))

fig_evolve = make_subplots(
    rows=2, cols=2,
    subplot_titles=[
        '💧 Yield Efficiency Index (Gen 1 = 1.0)',
        '🌫️ Minimum Operating RH (% — lower is better)',
        '🔋 Self-Healing Cycle Lifetime',
        '🌱 CO₂ Capture (tonnes/yr per unit)'
    ],
    vertical_spacing=0.15
)

gen_labels = [f"Gen {g} ({2026+g})" for g in generations]

metrics = [
    (yield_efficiency, MIRAG_GREEN, 'Yield Index', '💧', 1, 1),
    (rh_min_threshold, MIRAG_BLUE, 'Min RH %', '🌫️', 1, 2),
    ([c/1000 for c in self_healing_cycles], MIRAG_PURPLE, 'Cycles (×1000)', '🔋', 2, 1),
    (co2_per_unit, MIRAG_ORANGE, 'CO₂ t/yr', '🌱', 2, 2),
]

gen_x = [f"G{g}" for g in generations]

for data, color, name, icon, row, col in metrics:
    fig_evolve.add_trace(go.Scatter(
        x=gen_x, y=data,
        mode='lines+markers',
        name=f'{icon} {name}',
        line=dict(color=color, width=3),
        marker=dict(size=10, color=color, symbol='diamond',
                    line=dict(color='white', width=1)),
        fill='tozeroy', fillcolor=f'{color}15'
    ), row=row, col=col)

    # Add Gen 1 annotation
    if row == 1 and col == 1:
        fig_evolve.add_annotation(
            x='G1', y=data[0],
            text='Deploy 2027',
            showarrow=True, arrowcolor=color,
            font=dict(color=color, size=10, family='Courier New'),
            row=row, col=col
        )

fig_evolve.update_layout(
    height=600,
    title=dict(
        text='🧬 Innovation 19+22: Self-Evolving MIRAG™ — Each Generation Surpasses the Last',
        font=dict(color=MIRAG_GOLD, size=17, family='Courier New')
    ),
    **PLOTLY_DARK,
    showlegend=False
)

fig_evolve.show()

gen8_yield_gain = (yield_efficiency[-1] - 1) * 100
gen8_rh_min = rh_min_threshold[-1]
print(f"\n🧬 Self-Evolution Projection (2027 → 2034):")
print(f"   Yield efficiency: +{gen8_yield_gain:.0f}% by Generation 8")
print(f"   Minimum RH threshold: {gen8_rh_min:.1f}% (down from 15% at launch)")
print(f"   'Each unit deployed makes the next one better — no competitor can replicate this moat'")
```
