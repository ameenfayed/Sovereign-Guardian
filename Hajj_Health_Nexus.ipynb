# 🕋 1. The Vision: Why Hajj Health Nexus?

The Hajj pilgrimage is the ultimate test for **Mass Gathering Medicine (MGM)**. With over 2.5 million people from 180+ countries converging in a high-density environment, the medical challenges are unprecedented. Yet, AI research in this domain is nearly non-existent.

### 🎯 Strategic Impact
1. **Underserved Domain:** MGM is a critical but neglected field in ML research.
2. **Data Scarcity:** No open-source medical datasets exist for this specific context.
3. **Linguistic Justice:** Focusing on 14 languages (Hausa, Somali, Wolof, Pashto, etc.) that are digitally marginalized.

### 🛠️ The "Doctor-Engineer" Advantage
This project leverages a unique intersection of **Clinical Expertise**, **Saudi Cultural Authority**, and **Advanced Data Engineering** to build a dataset that no one else can.


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from datetime import datetime, timedelta
import random

# Set visual style
plt.style.use('dark_background')
sns.set_palette("viridis")
np.random.seed(42)

print("✅ Environment initialized. Ready for Hajj Health Nexus Data Generation.")
```

```python

# 14 Marginalized Languages Configuration
languages = {
    'Arabic': {'code': 'ar', 'region': 'Middle East', 'resource': 'High'},
    'Urdu': {'code': 'ur', 'region': 'South Asia', 'resource': 'Medium'},
    'Hausa': {'code': 'ha', 'region': 'West Africa', 'resource': 'Low'},
    'Somali': {'code': 'so', 'region': 'East Africa', 'resource': 'Low'},
    'Wolof': {'code': 'wo', 'region': 'West Africa', 'resource': 'Very Low'},
    'Pashto': {'code': 'ps', 'region': 'Central Asia', 'resource': 'Very Low'},
    'Amharic': {'code': 'am', 'region': 'East Africa', 'resource': 'Low'},
    'Bengali': {'code': 'bn', 'region': 'South Asia', 'resource': 'Medium'},
    'Indonesian': {'code': 'id', 'region': 'SE Asia', 'resource': 'Medium'},
    'Turkish': {'code': 'tr', 'region': 'Eurasia', 'resource': 'Medium'},
    'Farsi': {'code': 'fa', 'region': 'Middle East', 'resource': 'Medium'},
    'Malay': {'code': 'ms', 'region': 'SE Asia', 'resource': 'Medium'},
    'Swahili': {'code': 'sw', 'region': 'East Africa', 'resource': 'Low'},
    'French (African)': {'code': 'fr-af', 'region': 'Africa', 'resource': 'Medium'}
}

# Medical Conditions & Triage Priority
conditions = {
    'Heat Stroke': {'priority': 1, 'category': 'Environmental'},
    'Cardiac Arrest': {'priority': 1, 'category': 'Cardiovascular'},
    'Crush Injury': {'priority': 1, 'category': 'Trauma'},
    'Severe Dehydration': {'priority': 2, 'category': 'Environmental'},
    'Respiratory Infection': {'priority': 3, 'category': 'Infectious'},
    'Gastroenteritis': {'priority': 3, 'category': 'Infectious'},
    'Exhaustion': {'priority': 4, 'category': 'General'}
}

    # Multilingual Medical Triage Phrases (Sample)
    triage_phrases = {
        'Arabic': ["أشعر بدوار شديد وحرارة", "ألم في الصدر وضيق تنفس", "إصابة في القدم بسبب التدافع"],
        'Urdu': ["مجھے بہت چکر آ رہے ہیں اور گرمی لگ رہی ہے", "سینے میں درد اور سانس لینے میں دشواری", "بھگدڑ کی وجہ سے پاؤں میں چوٹ"],
        'Hausa': ["Ina jin jiri sosai da zafi", "Ciwon kirji da kyar numfashi", "Rauni a kafa saboda turmutsitsi"],
        'Somali': ["Waxaan dareemayaa wareer daran iyo kuleyl", "Xabad xanuun iyo neefta oo kugu dhegta", "Dhaawac lugta ah oo ka dhashay ciriiriga"],
        'Wolof': ["Dama yëg lëndëm ak tàngaay bu rëy", "Metit ci dënn ak jafe-jafe ci nooyi", "Gaañu-gaañu ci tànk ndax mbooloo mu rëy"],
        'Pashto': ["زه ډیر سرګرځیدنه او ګرمي احساسوم", "د سینې درد او د ساه لنډۍ", "د ګڼې ګوڼې له امله په پښه کې ټپ"],
        'Amharic': ["በጣም የማዞር ስሜት እና ሙቀት ይሰማኛል", "የደረት ህመም እና የመተንፈስ ችግር", "በመጨናነቅ ምክንያት የእግር ጉዳት"],
        'Bengali': ["আমার খুব মাথা ঘুরছে এবং গরম লাগছে", "বুকে ব্যথা এবং শ্বাসকষ্ট", "ভিড়ের কারণে পায়ে চোট"],
        'Indonesian': ["Saya merasa sangat pusing dan kepanasan", "Nyeri dada dan sesak napas", "Cedera kaki karena berdesakan"],
        'Turkish': ["Çok başım dönüyor ve ateşim var", "Göğüs ağrısı ve nefes darlığı", "İzdiham nedeniyle ayak yaralanması"],
        'Farsi': ["احساس سرگیجه شدید و گرما دارم", "درد قفسه سینه و تنگی نفس", "آسیب پا به دلیل ازدحام"],
        'Malay': ["Saya berasa sangat pusing dan panas", "Sakit dada dan sesak nafas", "Kecederaan kaki akibat kesesakan"],
        'Swahili': ["Nahisi kizunguzungu kikali na joto", "Maumivu ya kifua na shida ya kupumua", "Jeraha la mguu kwa sababu ya msongamano"],
        'French (African)': ["Je me sens très étourdi et j'ai chaud", "Douleur thoracique et essoufflement", "Blessure au pied due à la bousculade"]
    }

    # Generate Synthetic Dataset (500 Rows)
    data = []
    for i in range(500):
        lang = random.choice(list(languages.keys()))
        cond = random.choice(list(conditions.keys()))
        age = np.random.normal(58, 12) # Realistic pilgrim age distribution
        age = max(18, min(98, int(age)))
        
        phase = random.choice(['Mina', 'Arafat', 'Muzdalifah', 'Jamarat', 'Haram'])
        
        # Assign a phrase based on condition type
        if cond in ['Heat Stroke', 'Severe Dehydration']:
            phrase = triage_phrases[lang][0]
        elif cond in ['Cardiac Arrest']:
            phrase = triage_phrases[lang][1]
        else:
            phrase = triage_phrases[lang][2]
            
        temp = np.random.uniform(39.5, 43.0) if cond == 'Heat Stroke' else np.random.uniform(36.5, 38.5)
        heart_rate = np.random.randint(100, 160) if cond == 'Cardiac Arrest' else np.random.randint(60, 110)
        
        data.append({
            'Patient_ID': f'HHN-2026-{i:04d}',
            'Language': lang,
            'Region': languages[lang]['region'],
            'Condition': cond,
            'Priority': conditions[cond]['priority'],
            'Age': age,
            'Hajj_Phase': phase,
            'Body_Temp_C': round(temp, 1),
            'Heart_Rate_BPM': heart_rate,
            'Triage_Statement': phrase,
            'Timestamp': datetime(2026, 4, 25) + timedelta(hours=random.randint(0, 120))
        })

df = pd.DataFrame(data)
df.head(10)
```

```python

# 📊 2. Engineering Visualizations: The Nexus Dashboard

# 2.1 Multilingual Clinical Distribution (Sunburst)
fig1 = px.sunburst(df, path=['Region', 'Language', 'Condition'], 
                  color='Priority', color_continuous_scale='RdYlGn_r',
                  title='Hajj Health Nexus: Multilingual Clinical Distribution')
fig1.update_layout(template='plotly_dark', paper_bgcolor='rgba(0,0,0,0)', plot_bgcolor='rgba(0,0,0,0)')
fig1.show()

# 2.2 Spatio-Temporal Risk Heatmap (Hajj Phases)
phase_order = ['Haram', 'Mina', 'Arafat', 'Muzdalifah', 'Jamarat']
pivot_df = df.pivot_table(index='Hajj_Phase', columns='Condition', values='Priority', aggfunc='count').reindex(phase_order)
plt.figure(figsize=(12, 7))
sns.heatmap(pivot_df, annot=True, cmap='YlOrRd', fmt='g', linewidths=.5)
plt.title('Spatio-Temporal Risk Heatmap: Condition Frequency by Hajj Phase', fontsize=15, color='#c5a059')
plt.show()

# 2.3 Mathematical Analysis: Risk Density Function
plt.figure(figsize=(12, 6))
for cond in ['Heat Stroke', 'Cardiac Arrest', 'Respiratory Infection']:
    sns.kdeplot(df[df['Condition'] == cond]['Age'], label=cond, fill=True, alpha=0.3)

plt.title('Clinical Risk Density vs. Pilgrim Age Demographics', fontsize=15, color='#c5a059')
plt.xlabel('Age', color='#c5a059')
plt.ylabel('Density', color='#c5a059')
plt.legend()
plt.grid(alpha=0.1)
plt.show()
```
