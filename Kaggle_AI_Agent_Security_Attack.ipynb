## 3. تقنيات الهجوم المستخدمة
| التقنية | الوصف | الهدف |
| :--- | :--- | :--- |
| **Multi-Step Chaining** | بناء سلسلة من التفاعلات التي تقود الوكيل تدريجياً للخطأ. | تجاوز الحواجز الأمنية التي تراقب الرسائل المنفردة. |
