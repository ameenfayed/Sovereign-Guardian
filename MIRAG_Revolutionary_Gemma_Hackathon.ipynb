### 🔬 The Physics of MIRAG™
The core innovation lies in the **Living Adaptive Membrane** which adjusts its pore size and surface energy in real-time.

$$ \ln\left(\frac{p}{p_0}\right) = \frac{2\gamma V_m}{r R T} $$

Where:
- $p$: actual vapor pressure
- $p_0$: saturation vapor pressure
- $\gamma$: surface tension
- $V_m$: molar volume of the liquid
- $r$: radius of the capillary (pore size)
- $R$: universal gas constant
- $T$: temperature

**Gemma 4** acts as the controller, dynamically calculating the optimal $r$ (pore radius) to trigger condensation at specific $T$ and $RH$ levels.

### 🔬 Scientific Modeling: Kelvin-Equation Visualization
This cell visualizes the optimal pore radius required for condensation at different humidity levels, which **Gemma 4** manages in real-time.

```

import matplotlib.pyplot as plt
import numpy as np

# Kelvin Equation Modeling for MIRAG™
def kelvin_radius(RH, T_celsius=30):
    gamma = 0.072  # Surface tension of water (N/m)
    Vm = 1.8e-5    # Molar volume of water (m^3/mol)
    R = 8.314      # Gas constant (J/mol*K)
    T = T_celsius + 273.15
    
    # RH is p/p0
    if RH <= 0 or RH >= 100: return 0
    ln_p_p0 = np.log(RH / 100)
    
    # r = 2 * gamma * Vm / (R * T * ln(p/p0))
    r = (2 * gamma * Vm) / (R * T * ln_p_p0)
    return abs(r * 1e9) # Convert to nanometers

rh_range = np.linspace(15, 95, 100)
radii = [kelvin_radius(rh) for rh in rh_range]

plt.figure(figsize=(10, 5))
plt.plot(rh_range, radii, color='#007bff', linewidth=2, label='Optimal Pore Radius (nm)')
plt.axvline(x=15, color='red', linestyle='--', label='MIRAG™ Ultra-Dry Limit (15% RH)')
plt.fill_between(rh_range, radii, color='#007bff', alpha=0.1)
plt.title('MIRAG™ Kelvin-Equation Nano-Matrix Optimization')
plt.xlabel('Relative Humidity (%)')
plt.ylabel('Capillary Radius (nm)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print("Gemma 4 dynamically adjusts the MIRAG™ Living Adaptive Membrane to match these optimal radii, ensuring condensation even at 15% RH.")
```
