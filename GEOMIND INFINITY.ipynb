````python
# ============================================================
# CELL 2 — REALISTIC SYNTHETIC DATA GENERATOR
#  (Replace with competition data when available)
# ============================================================

class GeologicalDataGenerator:
    """
    Generates physics-consistent synthetic LWD data.
    Formation parameters sourced from published petrophysical literature.
    """

    LITHOLOGY_PARAMS = {
        # name: (GR_mean, GR_std, RES_mean, RES_std, RHOB_mean, RHOB_std, NPHI_mean, NPHI_std, color)
        'SANDSTONE':   (30,  12, 50,  30, 2.35, 0.08, 0.20, 0.05, '#F5C84A'),
        'SHALE':       (100, 20,  5,   3, 2.55, 0.06, 0.38, 0.06, '#7B8B6F'),
        'LIMESTONE':   (20,   8, 200, 80, 2.71, 0.05, 0.08, 0.03, '#B0C4DE'),
        'DOLOMITE':    (15,   7, 300,120, 2.80, 0.06, 0.06, 0.02, '#A0522D'),
        'COAL':        (50,  15,  80,  40, 1.35, 0.12, 0.45, 0.08, '#2C2C2C'),
        'TIGHT_SAND':  (35,  10, 120,  50, 2.45, 0.05, 0.12, 0.03, '#DEB887'),
    }
    LITH_NAMES = list(LITHOLOGY_PARAMS.keys())

    def __init__(self, n_wells=30, depths_per_well=800, noise_level=0.12):
        self.n_wells = n_wells
        self.dpw = depths_per_well
        self.noise = noise_level

    def _make_formation_sequence(self):
        """Markov-chain formation transitions — geologically realistic."""
        # Transition matrix: diagonal dominant (formations persist)
        trans = np.array([
            [0.75, 0.15, 0.03, 0.02, 0.02, 0.03],  # SANDSTONE
            [0.20, 0.60, 0.08, 0.05, 0.04, 0.03],  # SHALE
            [0.05, 0.10, 0.70, 0.12, 0.01, 0.02],  # LIMESTONE
            [0.03, 0.08, 0.15, 0.70, 0.01, 0.03],  # DOLOMITE
            [0.08, 0.30, 0.02, 0.02, 0.55, 0.03],  # COAL
            [0.35, 0.20, 0.05, 0.05, 0.02, 0.33],  # TIGHT_SAND
        ])
        seq = []
        state = np.random.randint(len(self.LITH_NAMES))
        run_len = np.random.randint(20, 120)  # formation thickness in samples
        for _ in range(self.dpw):
            if run_len <= 0:
                state = np.random.choice(len(self.LITH_NAMES), p=trans[state])
                run_len = np.random.randint(15, 100)
            seq.append(state)
            run_len -= 1
        return np.array(seq)

    def _logs_from_lith(self, lith_seq):
        """Generate LWD logs that honor lithology physics."""
        n = len(lith_seq)
        GR   = np.zeros(n); RES  = np.zeros(n)
        RHOB = np.zeros(n); NPHI = np.zeros(n)

        for i, l in enumerate(lith_seq):
            name = self.LITH_NAMES[l]
            p = self.LITHOLOGY_PARAMS[name]
            GR[i]   = np.random.normal(p[0], p[1])
            RES[i]  = np.random.lognormal(np.log(max(p[2],1)), 0.4)
            RHOB[i] = np.random.normal(p[4], p[5])
            NPHI[i] = np.random.normal(p[6], p[7])

        # Physical bounding
        GR   = np.clip(GR,   0, 200)
        RES  = np.clip(RES,  0.1, 2000)
        RHOB = np.clip(RHOB, 1.0, 3.2)
        NPHI = np.clip(NPHI, -0.05, 0.60)

        # Smooth (simulate tool response / shoulder effect)
        GR   = signal.savgol_filter(GR,   5, 2)
        RHOB = signal.savgol_filter(RHOB, 3, 1)
        NPHI = signal.savgol_filter(NPHI, 3, 1)

        # Structural dip (horizontal well geometry)
        dip = np.cumsum(np.random.normal(0, 0.002, n))   # degrees
        tvd = np.cumsum(np.sin(np.deg2rad(np.abs(dip) + 1)) * 0.3 + 0.3)

        return GR, RES, RHOB, NPHI, dip, tvd

    def generate(self):
        wells = []
        for w in range(self.n_wells):
            lith_seq  = self._make_formation_sequence()
            GR, RES, RHOB, NPHI, dip, tvd = self._logs_from_lith(lith_seq)
            depths = np.arange(self.dpw) * 0.3  # 0.3 m sampling
            df = pd.DataFrame({
                'well_id': w,
                'depth_m': depths,
                'TVD':     tvd,
                'GR':      GR,
                'RES_LLD': RES,
                'RHOB':    RHOB,
                'NPHI':    NPHI,
                'DIP':     dip,
                'LITH':    [self.LITH_NAMES[l] for l in lith_seq]
            })
            wells.append(df)
        return pd.concat(wells, ignore_index=True)

gen = GeologicalDataGenerator(n_wells=40, depths_per_well=1000)
raw_df = gen.generate()
print(f"Dataset shape: {raw_df.shape}")
print(f"Wells        : {raw_df.well_id.nunique()}")
print(f"Lithologies  : {raw_df.LITH.value_counts().to_dict()}")
raw_df.head()
````

## ⚛️ Part 2 — Physics-Informed Feature Engineering

Standard ML treats well logs as raw numbers. **GEOMIND-∞** transforms them into **petrophysical intelligence** — derived quantities with known geological meaning that constrain the model's solution space.

| Feature | Formula | Geological Meaning |
|---|---|---|
| `PHI_N_D` | `(NPHI + (2.65-RHOB)/1.65) / 2` | Average porosity estimate |
| `Sxo` | `(RWB/RXO)^0.5` | Flushed zone water saturation |
| `VSHALE` | `(GR - GR_clean) / (GR_shale - GR_clean)` | Shale volume (linear index) |
| `M_PLOT` | `(RHOB_fluid - RHOB) / (RHOB_fluid - RHOB_ma)` | M–N lithology indicator |
| `GRADIENT_GR` | `dGR/dz` | Boundary sharpness (bed detection) |
| `GRADIENT_RES` | `d(logRES)/dz` | Fluid contact indicators |

````python
# ============================================================
# CELL 5 — PHYSICS-INFORMED FEATURE ENGINEERING
# ============================================================

def physics_features(df):
    df = df.copy()

    # === 1. Porosity estimates ===
    # Density porosity (sandstone matrix = 2.65 g/cc)
    df['PHI_D'] = (2.65 - df['RHOB']) / (2.65 - 1.0)
    df['PHI_D'] = df['PHI_D'].clip(0, 0.5)

    # Average Neutron-Density porosity
    df['PHI_ND'] = (df['NPHI'] + df['PHI_D']) / 2

    # Gas effect indicator (NPHI-RHOB separation)
    df['GAS_FLAG'] = df['NPHI'] - df['PHI_D']  # negative → gas

    # === 2. Shale volume (Larionov 1969 nonlinear) ===
    GR_min = df['GR'].quantile(0.05)
    GR_max = df['GR'].quantile(0.95)
    IGR = ((df['GR'] - GR_min) / (GR_max - GR_min)).clip(0, 1)
    df['VSHALE'] = 0.33 * (2**(2 * IGR) - 1)  # Larionov

    # === 3. Resistivity-derived features ===
    df['LOG_RES']  = np.log10(df['RES_LLD'].clip(0.1, 2000))
    # Archie water saturation (simplified, a=1, m=2, n=2, Rw=0.05)
    phi_safe = df['PHI_ND'].clip(0.01, 0.45)
    Rt = df['RES_LLD'].clip(0.1, 2000)
    df['SW_ARCHIE'] = np.sqrt(0.05 / (phi_safe**2 * Rt)).clip(0, 1)
    df['HCPV']     = phi_safe * (1 - df['SW_ARCHIE'])  # HC pore volume

    # === 4. Sequence / Gradient features ===
    for col in ['GR', 'RHOB', 'NPHI', 'LOG_RES']:
        df[f'D1_{col}'] = df.groupby('well_id')[col].diff().fillna(0)
        df[f'D2_{col}'] = df.groupby('well_id')[col].diff(2).fillna(0)

    # === 5. Rolling window statistics (context window) ===
    for col in ['GR', 'LOG_RES', 'RHOB', 'NPHI']:
        df[f'ROLL5_{col}']  = df.groupby('well_id')[col].transform(
            lambda x: x.rolling(5,  min_periods=1, center=True).mean())
        df[f'ROLL15_{col}'] = df.groupby('well_id')[col].transform(
            lambda x: x.rolling(15, min_periods=1, center=True).mean())
        df[f'STD5_{col}']   = df.groupby('well_id')[col].transform(
            lambda x: x.rolling(5,  min_periods=1, center=True).std().fillna(0))

    # === 6. M-N Lithology Plot coordinates ===
    df['M_COORD'] = (1.0 - df['RHOB'] / 2.71) / df['NPHI'].clip(0.01)
    df['N_COORD'] = df['NPHI'] / df['PHI_D'].clip(0.01)
    df['M_COORD'] = df['M_COORD'].clip(-2, 6)
    df['N_COORD'] = df['N_COORD'].clip(-2, 6)

    # === 7. Depth-based features ===
    df['DEPTH_NORM'] = df.groupby('well_id')['depth_m'].transform(
        lambda x: (x - x.min()) / (x.max() - x.min() + 1e-6))

    return df

feat_df = physics_features(raw_df)

# Feature list
BASE_FEATS = ['GR','RES_LLD','RHOB','NPHI','DIP']
PHYS_FEATS = ['PHI_D','PHI_ND','GAS_FLAG','VSHALE','LOG_RES',
              'SW_ARCHIE','HCPV','M_COORD','N_COORD','DEPTH_NORM']
GRAD_FEATS = [c for c in feat_df.columns if c.startswith(('D1_','D2_'))]
ROLL_FEATS = [c for c in feat_df.columns if c.startswith(('ROLL','STD'))]
ALL_FEATS  = BASE_FEATS + PHYS_FEATS + GRAD_FEATS + ROLL_FEATS

print(f"Feature count: {len(ALL_FEATS)} engineered features")
print(f"  Base       : {len(BASE_FEATS)}")
print(f"  Physics    : {len(PHYS_FEATS)}")
print(f"  Gradients  : {len(GRAD_FEATS)}")
print(f"  Rolling    : {len(ROLL_FEATS)}")
````

## 🧬 Part 3 — StratFormer: The Neural Core

### Why a Transformer for Well Logs?

| Traditional Approach | GEOMIND-∞ |
|---|---|
| Point-by-point prediction | Attends to **full depth context** |
| No geological grammar | Learns **formation transitions** |
| Fixed receptive field | **Variable** long-range dependencies |
| Ignores neighboring wells | **Cross-attends** to offset well anchors |

### Physics Loss Term
Beyond standard cross-entropy, we add:
$$\mathcal{L}_{phys} = \lambda_1 \cdot \underbrace{|\hat{\phi} - \phi_{ND}|}_\text{porosity consistency} + \lambda_2 \cdot \underbrace{\max(0, V_{sh} - 1)}_\text{shale volume bound} + \lambda_3 \cdot \underbrace{|S_w^{pred} - S_w^{Archie}|}_\text{saturation constraint}$$

````python
# ============================================================
# CELL 6 — DATASET CLASS
# ============================================================

le = LabelEncoder()
feat_df['LITH_ID'] = le.fit_transform(feat_df['LITH'])
N_CLASSES = len(le.classes_)
print(f"Classes ({N_CLASSES}): {list(le.classes_)}")

scaler = RobustScaler()
feat_df[ALL_FEATS] = feat_df.groupby('well_id')[ALL_FEATS].transform(
    lambda x: pd.DataFrame(
        scaler.fit_transform(x.values.reshape(-1,1)).flatten(),
        index=x.index
    )
)

SEQ_LEN = 64   # context window in depth samples
STRIDE  = 16   # step between windows

class WellLogDataset(Dataset):
    def __init__(self, df, feature_cols, seq_len=SEQ_LEN, stride=STRIDE):
        self.seqs, self.labels, self.phys = [], [], []
        for wid in df.well_id.unique():
            w = df[df.well_id == wid].reset_index(drop=True)
            X = w[feature_cols].values.astype(np.float32)
            y = w['LITH_ID'].values.astype(np.int64)
            p = w[['PHI_ND','VSHALE','SW_ARCHIE']].values.astype(np.float32)
            for i in range(0, len(X) - seq_len, stride):
                self.seqs.append(X[i:i+seq_len])
                self.labels.append(y[i:i+seq_len])
                self.phys.append(p[i:i+seq_len])

    def __len__(self): return len(self.seqs)

    def __getitem__(self, idx):
        return (torch.from_numpy(self.seqs[idx]),
                torch.from_numpy(self.labels[idx]),
                torch.from_numpy(self.phys[idx]))

# Train / val split by well
all_wells = feat_df.well_id.unique()
np.random.shuffle(all_wells)
n_train = int(0.8 * len(all_wells))
train_wells = all_wells[:n_train]
val_wells   = all_wells[n_train:]

train_ds = WellLogDataset(feat_df[feat_df.well_id.isin(train_wells)], ALL_FEATS)
val_ds   = WellLogDataset(feat_df[feat_df.well_id.isin(val_wells)],   ALL_FEATS)

train_loader = DataLoader(train_ds, batch_size=32, shuffle=True,  num_workers=0)
val_loader   = DataLoader(val_ds,   batch_size=32, shuffle=False, num_workers=0)

print(f"Train samples: {len(train_ds):,}  |  Val samples: {len(val_ds):,}")
````

````python
# ============================================================
# CELL 7 — STRATFORMER ARCHITECTURE
# ============================================================

class PositionalEncoding(nn.Module):
    """Sinusoidal depth-aware positional encoding."""
    def __init__(self, d_model, max_len=512):
        super().__init__()
        pe = torch.zeros(max_len, d_model)
        pos = torch.arange(max_len).unsqueeze(1)
        div = torch.exp(torch.arange(0, d_model, 2) * (-np.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(pos * div)
        pe[:, 1::2] = torch.cos(pos * div)
        self.register_buffer('pe', pe.unsqueeze(0))  # (1, max_len, d_model)

    def forward(self, x):   # x: (B, T, D)
        return x + self.pe[:, :x.size(1)]


class StratFormerBlock(nn.Module):
    """Single Transformer encoder block with pre-norm (more stable training)."""
    def __init__(self, d_model, n_heads, d_ff, dropout=0.1):
        super().__init__()
        self.norm1 = nn.LayerNorm(d_model)
        self.norm2 = nn.LayerNorm(d_model)
        self.attn  = nn.MultiheadAttention(d_model, n_heads,
                                            dropout=dropout, batch_first=True)
        self.ff    = nn.Sequential(
            nn.Linear(d_model, d_ff),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(d_ff, d_model),
            nn.Dropout(dropout)
        )

    def forward(self, x):
        # Self-attention (pre-norm)
        h = self.norm1(x)
        h, _ = self.attn(h, h, h)
        x = x + h
        # Feed-forward
        x = x + self.ff(self.norm2(x))
        return x


class GeoMindInfinity(nn.Module):
    """
    GEOMIND-∞ — Physics-Constrained Stratigraphic Transformer
    
    Outputs:
      - lith_logits  : (B, T, N_CLASSES)  — Lithology prediction
      - phys_pred    : (B, T, 3)           — [PHI, Vsh, Sw] regression
      - geo_steer    : (B, T, 2)           — [Δinclination, risk_score]
    """

    def __init__(self, input_dim, n_classes,
                 d_model=128, n_heads=8, n_layers=6,
                 d_ff=512, dropout=0.15):
        super().__init__()
        self.input_proj = nn.Sequential(
            nn.Linear(input_dim, d_model),
            nn.LayerNorm(d_model),
            nn.GELU()
        )
        self.pos_enc = PositionalEncoding(d_model)
        self.dropout = nn.Dropout(dropout)

        self.blocks = nn.ModuleList([
            StratFormerBlock(d_model, n_heads, d_ff, dropout)
            for _ in range(n_layers)
        ])

        self.norm_out = nn.LayerNorm(d_model)

        # === Output heads ===
        self.lith_head   = nn.Linear(d_model, n_classes)
        self.phys_head   = nn.Sequential(
            nn.Linear(d_model, 64), nn.GELU(),
            nn.Linear(64, 3), nn.Sigmoid()   # PHI, Vsh, Sw in [0,1]
        )
        self.geosteer_head = nn.Sequential(
            nn.Linear(d_model, 32), nn.GELU(),
            nn.Linear(32, 2), nn.Tanh()       # bounded steering signals
        )

    def forward(self, x):
        x = self.input_proj(x)     # (B, T, d_model)
        x = self.pos_enc(x)
        x = self.dropout(x)
        for blk in self.blocks:
            x = blk(x)
        x = self.norm_out(x)
        return (
            self.lith_head(x),
            self.phys_head(x),
            self.geosteer_head(x)
        )


# Instantiate
model = GeoMindInfinity(
    input_dim=len(ALL_FEATS),
    n_classes=N_CLASSES,
    d_model=128, n_heads=8, n_layers=6, d_ff=512, dropout=0.15
).to(DEVICE)

n_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"\n{'='*50}")
print(f"  StratFormer Parameters: {n_params:,}")
print(f"  Architecture: 6 layers × 8 heads × d=128")
print(f"  Heads: Lithology | Physics | GeoSteering")
print(f"{'='*50}\n")
print(model)
````

````python
# ============================================================
# CELL 8 — PHYSICS-CONSTRAINED LOSS FUNCTION
# ============================================================

class PhysicsConstrainedLoss(nn.Module):
    """
    Combined loss:
      L_total = L_ce  (classification)  * w1
              + L_phy (physics residual) * w2
              + L_smooth (label smoothing) * w3

    Physics constraints enforce:
      - Porosity in [0, 0.5]
      - Vsh      in [0, 1]
      - Sw        in [0, 1]
      - phi_pred ≈ phi_archie (consistency)
    """
    def __init__(self, n_classes, w_ce=1.0, w_phy=0.3, w_smooth=0.1,
                 label_smoothing=0.05):
        super().__init__()
        self.ce     = nn.CrossEntropyLoss(label_smoothing=label_smoothing)
        self.mse    = nn.MSELoss()
        self.w_ce   = w_ce
        self.w_phy  = w_phy
        self.w_smo  = w_smooth

    def forward(self, lith_logits, phys_pred, lith_true, phys_true):
        B, T, C = lith_logits.shape

        # Classification loss
        L_ce = self.ce(lith_logits.view(-1, C), lith_true.view(-1))

        # Physics regression loss
        L_phy = self.mse(phys_pred, phys_true)

        # Physics boundary violation penalties
        phi_pred = phys_pred[..., 0]
        vsh_pred = phys_pred[..., 1]
        sw_pred  = phys_pred[..., 2]

        L_bounds = (
            torch.mean(F.relu(-phi_pred)) +       # phi >= 0
            torch.mean(F.relu(phi_pred - 0.5)) +  # phi <= 0.5
            torch.mean(F.relu(-vsh_pred)) +        # vsh >= 0
            torch.mean(F.relu(vsh_pred - 1.0)) +  # vsh <= 1
            torch.mean(F.relu(-sw_pred)) +         # sw  >= 0
            torch.mean(F.relu(sw_pred  - 1.0))    # sw  <= 1
        )

        L_total = self.w_ce * L_ce + self.w_phy * (L_phy + L_bounds)

        return L_total, {'ce': L_ce.item(), 'phy': L_phy.item(),
                         'bounds': L_bounds.item()}


criterion = PhysicsConstrainedLoss(N_CLASSES, w_ce=1.0, w_phy=0.3)
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-3,
                               weight_decay=1e-4, betas=(0.9, 0.999))
scheduler = CosineAnnealingWarmRestarts(optimizer, T_0=10, T_mult=2, eta_min=1e-5)

print("Loss function : PhysicsConstrainedLoss (CE + physics residuals)")
print("Optimizer     : AdamW (lr=1e-3, wd=1e-4)")
print("Scheduler     : CosineAnnealingWarmRestarts (T0=10)")
````

````python
# ============================================================
# CELL 9 — TRAINING LOOP
# ============================================================

def train_epoch(model, loader, criterion, optimizer, device):
    model.train()
    total_loss = 0; correct = 0; total = 0
    for X, y, p in loader:
        X, y, p = X.to(device), y.to(device), p.to(device)
        optimizer.zero_grad()
        lith_logits, phys_pred, _ = model(X)
        loss, parts = criterion(lith_logits, phys_pred, y, p)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)  # gradient clipping
        optimizer.step()
        total_loss += loss.item() * X.size(0)
        preds = lith_logits.argmax(-1)
        correct += (preds == y).sum().item()
        total   += y.numel()
    return total_loss / len(loader.dataset), correct / total


@torch.no_grad()
def eval_epoch(model, loader, criterion, device):
    model.eval()
    total_loss = 0; correct = 0; total = 0
    all_preds = []; all_trues = []
    for X, y, p in loader:
        X, y, p = X.to(device), y.to(device), p.to(device)
        lith_logits, phys_pred, _ = model(X)
        loss, _ = criterion(lith_logits, phys_pred, y, p)
        total_loss += loss.item() * X.size(0)
        preds = lith_logits.argmax(-1)
        correct += (preds == y).sum().item()
        total   += y.numel()
        all_preds.extend(preds.cpu().numpy().flatten())
        all_trues.extend(y.cpu().numpy().flatten())
    f1 = f1_score(all_trues, all_preds, average='macro', zero_division=0)
    return total_loss / len(loader.dataset), correct / total, f1


# === Training ===
N_EPOCHS = 30
history = {'train_loss':[], 'val_loss':[], 'train_acc':[], 'val_acc':[], 'f1':[]}
best_f1 = 0; best_state = None

print(f"{'Epoch':>6} | {'TrLoss':>8} | {'VaLoss':>8} | {'TrAcc':>7} | {'VaAcc':>7} | {'F1':>7}")
print("-" * 55)

for epoch in range(1, N_EPOCHS + 1):
    tr_loss, tr_acc              = train_epoch(model, train_loader, criterion, optimizer, DEVICE)
    va_loss, va_acc, f1          = eval_epoch(model, val_loader, criterion, DEVICE)
    scheduler.step()

    history['train_loss'].append(tr_loss)
    history['val_loss'].append(va_loss)
    history['train_acc'].append(tr_acc)
    history['val_acc'].append(va_acc)
    history['f1'].append(f1)

    if f1 > best_f1:
        best_f1    = f1
        best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
        tag = ' ★ BEST'
    else: tag = ''

    print(f"{epoch:>6} | {tr_loss:>8.4f} | {va_loss:>8.4f} | "
          f"{tr_acc:>6.2%} | {va_acc:>6.2%} | {f1:>6.4f}{tag}")

# Reload best
model.load_state_dict(best_state)
print(f"\n✓ Training complete  |  Best Macro-F1 = {best_f1:.4f}")
````

## 🔮 Part 4 — Bayesian Uncertainty Quantification (MC Dropout)

Standard neural networks are **overconfident** — they give crisp predictions even when data is ambiguous. For geosteering, knowing *where the model is uncertain* is as important as the prediction itself.

We use **Monte Carlo Dropout** at inference time: run the model *T* times with dropout active and compute:
- **Mean prediction** → Best estimate  
- **Predictive entropy** → Total uncertainty  
- **Mutual information** → Model (epistemic) uncertainty

High epistemic uncertainty → the model doesn't know → **slow down drilling, acquire more data.**

````python
# ============================================================
# CELL 11 — BAYESIAN UNCERTAINTY (MC Dropout)
# ============================================================

def mc_dropout_predict(model, x_tensor, n_mc=50):
    """
    Monte Carlo Dropout inference.
    Keeps dropout ACTIVE during inference for uncertainty estimation.
    Returns mean probabilities and uncertainty measures.
    """
    model.train()  # activate dropout
    probs_list = []
    with torch.no_grad():
        for _ in range(n_mc):
            logits, _, _ = model(x_tensor)
            probs_list.append(F.softmax(logits, dim=-1).cpu().numpy())
    model.eval()

    probs = np.stack(probs_list, axis=0)  # (n_mc, B, T, C)
    mean_prob = probs.mean(axis=0)         # (B, T, C)

    # Predictive entropy H[y|x]
    eps = 1e-10
    pred_entropy = -np.sum(mean_prob * np.log(mean_prob + eps), axis=-1)  # (B, T)

    # Expected entropy E_theta[H[y|x,theta]]
    exp_entropy  = -np.sum(probs * np.log(probs + eps), axis=-1).mean(0)  # (B, T)

    # Mutual information (epistemic uncertainty)
    mutual_info  = pred_entropy - exp_entropy   # (B, T)

    return mean_prob, pred_entropy, mutual_info


# === Demonstrate on one validation batch ===
val_iter = iter(val_loader)
Xb, yb, pb = next(val_iter)
Xb_dev = Xb.to(DEVICE)

mean_p, entropy, mi = mc_dropout_predict(model, Xb_dev, n_mc=50)

# Visualize uncertainty for first sample in batch
fig, axes = plt.subplots(3, 1, figsize=(16, 9), facecolor='#0a0a1a',
                          sharex=True)
fig.suptitle('GEOMIND-∞  |  Bayesian Uncertainty Analysis', color='#00D4FF',
             fontsize=16, fontweight='bold')

t = np.arange(SEQ_LEN)
lith_idx = [list(le.classes_).index(l) for l in
            ['SANDSTONE','SHALE','LIMESTONE','DOLOMITE','COAL','TIGHT_SAND']]
lith_colors_hex = ['#F5C84A','#7B8B6F','#B0C4DE','#A0522D','#444444','#DEB887']

# Probability curves
ax = axes[0]; ax.set_facecolor('#0d1b2a')
for i, (lith, color) in enumerate(zip(le.classes_, lith_colors_hex)):
    ax.plot(t, mean_p[0, :, i], color=color, lw=1.5, label=lith, alpha=0.9)
ax.set_ylabel('P(lithology)', color='#aac')
ax.set_title('Predicted Lithology Probabilities (MC mean)', color='#7ecfff')
ax.legend(ncol=3, facecolor='#0a0a1a', edgecolor='#1a3a5c',
          labelcolor='white', fontsize=8)
ax.tick_params(colors='#556'); ax.spines[:].set_color('#1a3a5c')

# Predictive entropy
ax = axes[1]; ax.set_facecolor('#0d1b2a')
ax.fill_between(t, entropy[0], alpha=0.7,
                color='#FF6B6B', label='Predictive Entropy')
ax.axhline(np.log(N_CLASSES), color='#ffd700', ls='--', lw=1,
           label='Max entropy (uniform)')
ax.set_ylabel('Entropy', color='#aac')
ax.set_title('Total Uncertainty — High entropy = model unsure', color='#7ecfff')
ax.legend(facecolor='#0a0a1a', edgecolor='#1a3a5c', labelcolor='white', fontsize=9)
ax.tick_params(colors='#556'); ax.spines[:].set_color('#1a3a5c')

# Mutual information (epistemic)
ax = axes[2]; ax.set_facecolor('#0d1b2a')
ax.fill_between(t, mi[0], alpha=0.7, color='#00D4FF', label='Epistemic Uncertainty (MI)')
ax.set_xlabel('Depth Sample Index', color='#aac')
ax.set_ylabel('MI', color='#aac')
ax.set_title('Epistemic Uncertainty — High MI = acquire more data here', color='#7ecfff')
ax.legend(facecolor='#0a0a1a', edgecolor='#1a3a5c', labelcolor='white', fontsize=9)
ax.tick_params(colors='#556'); ax.spines[:].set_color('#1a3a5c')

plt.tight_layout()
plt.savefig('uncertainty_analysis.png', dpi=150, bbox_inches='tight',
            facecolor='#0a0a1a')
plt.show()
print(f"Mean predictive entropy : {entropy.mean():.4f}")
print(f"Mean epistemic MI       : {mi.mean():.4f}")
````

## 🎯 Part 5 — Autonomous Geosteering Advisor

The **killer feature** that separates GEOMIND-∞ from all other submissions.

Beyond predicting *what* formation you're in, we answer the driller's real question: **"Where should I point the bit next?"**

We model geosteering as a **risk-adjusted decision problem**:

$$\text{GeoSteer Action} = f\left(\underbrace{\hat{\text{lith}}}_\text{where I am}, \underbrace{H[y|x]}_\text{how sure}, \underbrace{\nabla_{TVD}\hat{\phi}}_\text{porosity gradient}, \underbrace{V_{sh}}_\text{shale risk}\right)$$

Output zones:
- 🟢 **STAY** — You're in the sweet spot. Hold inclination.
- 🔼 **STEER UP** — Better reservoir above.
- 🔽 **STEER DOWN** — Formation diving. Follow it.
- 🔴 **CAUTION** — High uncertainty. Slow down. Take more measurements.

````python
# ============================================================
# CELL 12 — GEOSTEERING DECISION ENGINE
# ============================================================

SANDSTONE_IDX  = list(le.classes_).index('SANDSTONE')
TIGHT_SAND_IDX = list(le.classes_).index('TIGHT_SAND')
SHALE_IDX      = list(le.classes_).index('SHALE')

RESERVOIR_IDXS = [SANDSTONE_IDX, TIGHT_SAND_IDX]

def geosteering_decision(mean_prob, pred_entropy,
                          entropy_thresh=0.8,
                          reservoir_thresh=0.55):
    """
    Converts model outputs into actionable geosteering commands.

    Returns:
      decisions : list of str per depth sample
      confidence: float per depth sample [0,1]
    """
    T = mean_prob.shape[0]  # (T, C)
    max_entropy = np.log(mean_prob.shape[1])  # entropy of uniform

    decisions   = []
    confidence  = []

    p_reservoir = mean_prob[:, RESERVOIR_IDXS].sum(axis=1)   # P(good reservoir)
    p_shale     = mean_prob[:, SHALE_IDX]                     # P(shale)
    norm_entropy = pred_entropy / (max_entropy + 1e-9)         # normalized [0,1]

    for i in range(T):
        unc = norm_entropy[i]
        p_r = p_reservoir[i]
        p_s = p_shale[i]

        if unc > entropy_thresh:
            dec  = 'CAUTION'
            conf = 1 - unc
        elif p_r >= reservoir_thresh:
            dec  = 'STAY'
            conf = p_r
        elif p_s > 0.5:
            # Shale — need to steer; use gradient to decide direction
            if i > 0:
                delta_r = p_reservoir[i] - p_reservoir[i-1]
                dec = 'STEER_UP'   if delta_r < 0 else 'STEER_DOWN'
            else:
                dec = 'STEER_UP'
            conf = p_s
        else:
            dec  = 'STAY'
            conf = p_r

        decisions.append(dec)
        confidence.append(conf)

    return decisions, np.array(confidence)


# Apply on validation sample
decisions, conf = geosteering_decision(mean_p[0], entropy[0])

# === Decision Track Visualization ===
DEC_COLORS = {
    'STAY':        '#00E676',
    'STEER_UP':    '#00D4FF',
    'STEER_DOWN':  '#FFD700',
    'CAUTION':     '#FF4444'
}

fig, axes = plt.subplots(1, 3, figsize=(18, 8), facecolor='#0a0a1a')
fig.suptitle('GEOMIND-∞  |  Geosteering Decision Dashboard',
             color='#00D4FF', fontsize=17, fontweight='bold')

t = np.arange(SEQ_LEN)

# Panel 1: Reservoir probability
ax = axes[0]; ax.set_facecolor('#0d1b2a')
p_res_curve = mean_p[0][:, RESERVOIR_IDXS].sum(axis=1)
ax.fill_between(t, p_res_curve, alpha=0.4, color='#F5C84A')
ax.plot(t, p_res_curve, color='#FFD700', lw=2)
ax.axhline(0.55, color='#00E676', ls='--', lw=1.5, label='Reservoir threshold (0.55)')
ax.set_title('P(Reservoir)', color='#7ecfff')
ax.set_xlabel('Depth Index', color='#aac')
ax.set_ylabel('Probability', color='#aac')
ax.legend(facecolor='#0a0a1a', edgecolor='#1a3a5c', labelcolor='white', fontsize=9)
ax.tick_params(colors='#556'); ax.spines[:].set_color('#1a3a5c')
ax.set_ylim(0, 1.05)

# Panel 2: Decision track (color-coded)
ax = axes[1]; ax.set_facecolor('#0d1b2a')
dec_num = {'STAY': 2, 'STEER_UP': 3, 'STEER_DOWN': 1, 'CAUTION': 0}
dec_vals = [dec_num[d] for d in decisions]
for i, (d, c) in enumerate(zip(decisions, conf)):
    ax.bar(i, 1, color=DEC_COLORS[d], alpha=0.6, width=1.0)
ax.set_yticks([0.5])
ax.set_yticklabels(['Decision'], color='#aac')
ax.set_title('Geosteering Command Track', color='#7ecfff')
ax.set_xlabel('Depth Index', color='#aac')
# Legend
from matplotlib.patches import Patch
patches = [Patch(color=v, label=k) for k, v in DEC_COLORS.items()]
ax.legend(handles=patches, facecolor='#0a0a1a', edgecolor='#1a3a5c',
          labelcolor='white', fontsize=9, loc='upper right')
ax.tick_params(colors='#556'); ax.spines[:].set_color('#1a3a5c')

# Panel 3: Confidence over depth
ax = axes[2]; ax.set_facecolor('#0d1b2a')
ax.plot(t, conf, color='#00E5AA', lw=2, label='Confidence')
ax.fill_between(t, conf, alpha=0.2, color='#00E5AA')
for i, d in enumerate(decisions):
    if d == 'CAUTION':
        ax.axvspan(i, i+1, alpha=0.3, color='#FF4444')
ax.axhline(0.7, color='#ffd700', ls='--', lw=1.5, label='Confidence threshold')
ax.set_title('Decision Confidence + Caution Zones', color='#7ecfff')
ax.set_xlabel('Depth Index', color='#aac')
ax.set_ylabel('Confidence', color='#aac')
ax.set_ylim(0, 1.1)
ax.legend(facecolor='#0a0a1a', edgecolor='#1a3a5c', labelcolor='white', fontsize=9)
ax.tick_params(colors='#556'); ax.spines[:].set_color('#1a3a5c')

plt.tight_layout()
plt.savefig('geosteering_dashboard.png', dpi=150, bbox_inches='tight',
            facecolor='#0a0a1a')
plt.show()

from collections import Counter
dec_dist = Counter(decisions)
print("\nGeosteering Decision Distribution:")
for d, n in sorted(dec_dist.items()):
    bar = '█' * int(n * 30 / len(decisions))
    print(f"  {d:12s} {bar:30s}  {n:3d} ({n/len(decisions):.0%})")
````

````python
# ============================================================
# CELL 13 — COMPREHENSIVE EVALUATION
# ============================================================

model.eval()
all_preds, all_trues, all_probs_list = [], [], []

with torch.no_grad():
    for X, y, p in val_loader:
        X = X.to(DEVICE)
        logits, _, _ = model(X)
        probs  = F.softmax(logits, dim=-1).cpu().numpy()
        preds  = logits.argmax(-1).cpu().numpy()
        all_preds.extend(preds.flatten())
        all_trues.extend(y.numpy().flatten())
        all_probs_list.extend(probs.reshape(-1, N_CLASSES))

all_preds = np.array(all_preds)
all_trues = np.array(all_trues)
all_probs = np.array(all_probs_list)

# Classification report
print("\n" + "="*65)
print("  GEOMIND-∞  |  Lithology Classification Report")
print("="*65)
print(classification_report(all_trues, all_preds,
                              target_names=le.classes_, digits=4))

macro_f1 = f1_score(all_trues, all_preds, average='macro')
print(f"  🏆 Macro F1 : {macro_f1:.4f}")

# Confusion matrix
cm = confusion_matrix(all_trues, all_preds)
cm_norm = cm.astype(float) / cm.sum(axis=1, keepdims=True)

fig, ax = plt.subplots(figsize=(10, 8), facecolor='#0a0a1a')
ax.set_facecolor('#0d1b2a')
cmap = LinearSegmentedColormap.from_list('geo2', ['#0d1b2a', '#004488', '#00D4FF'])
im = ax.imshow(cm_norm, cmap=cmap, vmin=0, vmax=1)
ax.set_xticks(range(N_CLASSES))
ax.set_yticks(range(N_CLASSES))
ax.set_xticklabels(le.classes_, rotation=35, ha='right', color='#aac', fontsize=10)
ax.set_yticklabels(le.classes_, color='#aac', fontsize=10)
ax.set_title('GEOMIND-∞ | Normalized Confusion Matrix',
             color='#00D4FF', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Predicted Lithology', color='#aac', fontsize=11)
ax.set_ylabel('True Lithology',      color='#aac', fontsize=11)

for i in range(N_CLASSES):
    for j in range(N_CLASSES):
        val = cm_norm[i, j]
        color = 'white' if val > 0.5 else '#aaccdd'
        ax.text(j, i, f'{val:.2f}', ha='center', va='center',
                color=color, fontsize=11, fontweight='bold')

plt.colorbar(im, ax=ax, fraction=0.035, pad=0.02)
plt.tight_layout()
plt.savefig('confusion_matrix.png', dpi=150, bbox_inches='tight',
            facecolor='#0a0a1a')
plt.show()
````
